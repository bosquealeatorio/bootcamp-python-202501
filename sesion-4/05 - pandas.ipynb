## Chaining

````python
pdf_athletes = pd.read_csv("athlete_events.csv")
````

````python
_pdf = (
    pdf_athletes
    .assign(Game = pdf_athletes["City"] + " " + pdf_athletes["Year"].astype(str))
    [["Game", "NOC", "Team", "Event", "Medal"]]
    .dropna(subset=["Medal"])
    .drop_duplicates()
    .groupby(by = ["Game", "NOC", "Team"], as_index=False)
    .count()
    .sort_values(by=["Medal", "Event"], ascending=[False, False])
)
````

````python
_pdf.head()
````

````python
_pdf.to_excel("medallero_olimpico_por_juego.xlsx", index=False)
````
