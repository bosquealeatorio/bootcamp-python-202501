## While

```
while condition:
    
    code to execute
```

while's anatomy

````python
x = 0

while x < 5:
    print('Not there yet ... x = ', x)
    x = x + 1

print("Final value of x = ", x)
````

pitfalls

````python
#sum of numbers < 10
x = 1
sum = 0
while x < 10:
    sum += x #sum = sum + x
    x += 1


#product of numbers < 10
x = 1
product = 1
while x < 10:
    product *= x #product = product * x
    x += 1

print(sum, product)
````

infinite loops

````python
x =1 
sum = 0

while x < 10:
    sum += x
    print(sum)
````

`break`

````python
x = 1

while True:
    print(x)
    x += 1

    if x > 10:
        break
````

````python
lucky_number = 22

def guess_my_lucky_number():
    """ask user for a number between 1 and 100 
    and stops when user guess the lucky number"""

    while True:
        
        number = int(input("Guess a number between 1 and 10: "))
        
        if number == lucky_number:
            print('You guessed it!')
            break
            
        else:
            print('Try again!')
````

Calculate the average of the first 20 even numbers

````python
sum = 0
count = 0

for number in range(2,42,2):
    sum += number
    count += 1

print("Average: ", sum/count)
````

nesting

````python
for a in range(5):
    for b in range(5, 10):
        print(a, b)
````

print all the domino pieces

````python
for i in range(7):
    for j in range(7):
        if j>=i:
            print("*"*i, '|', '*'*j)
````

````python
for i in range(7):
    for j in range(i, 7):
        print("*"*i + ' | ' + '*'*j)
````
